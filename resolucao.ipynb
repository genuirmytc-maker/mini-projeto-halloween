## BLOCO 2 LIMPEZA DE STRINGS COM REGEX

```python
def limpar_categoria(texto):
    if pd.isna(texto) or str(texto).strip() == '':
        return 'sem categoria'
    
    # Aplica strip() para remover espaços extras e lower() para minúsculas
    texto_limpo = str(texto).strip().lower()
    
    # re.sub() para retirar pontuações, mantendo apenas letras sem acento, números e espaços
    texto_limpo = re.sub(r'[^a-z0-9\s]', '', texto_limpo)
    
    # Substitui múltiplos espaços por apenas um espaço simples
    texto_limpo = re.sub(r'\s+', ' ', texto_limpo).strip()
    
    if texto_limpo == '':
        return 'sem categoria'
    
    return texto_limpo

# Aplicamos a função de limpeza na coluna categoria
df['categoria'] = df['categoria'].apply(limpar_categoria)
```

# 3. Função para padronizar os canais de venda.

```python
def padronizar_canal(canal):
    if pd.isna(canal):
        return 'Desconhecido'
    c = str(canal) .strip().lower()
    if 'mercado' in c:
        return 'Mercado'
    elif 'loja' in c or 'física' in c or 'fisica' in c:
        return 'Loja física'
    elif 'web' in c or 'site' in c:
        return 'Web'
    return 'Desconhecido'
# Aplicamos a padronização na coluna canal_venda
df['canal_venda'] = df['canal_venda'].apply(padronizar_canal)

```

# BLOCO 3  REGRA E VALIDAÇÃO DE ENTREGAS

```python
# 1. Identificamos registros sem data_entrega (nulos ou em branco)

sem_entrega = df['data_entrega'].isna() | (df['data_entrega'].astype(str).str.strip() == '')

# 2. Grupo 1: Sem entrega E com status Cancelado

sem_entrega_cancelados = sem_entrega & (df['status_pedido'].astype(str).str.strip() == 'Cancelado')

# 3. Grupo 2: Sem entrega E com OUTRO status (Contraexemplos)

sem_entrega_outros = sem_entrega & (df['status_pedido'].astype(str).str.strip() != 'Cancelado')

print("=== BLOCO 3: TESTE DE HIPÓTESE DE ENTREGAS ===")
print(f"1. Pedidos sem entrega e Cancelados: {sem_entrega_cancelados.sum()}")
print(f"2. Pedidos sem entrega com OUTROS status: {sem_entrega_outros.sum()}")

# 4. Avaliação da hipótese
if sem_entrega_outros.sum() > 0:
    print("\nResultado: A hipótese é FALSA! Existem pedidos sem data de entrega que NÃO foram cancelados.")
    print("\nExibindo até 3 exemplos do segundo grupo (contraexemplos):")
    display(df[sem_entrega_outros][['pedido_id', 'status_pedido', 'data_venda', 'data_entrega']].head(3))
else:
    print("\nResultado: A hipótese é VERDADEIRA! Todos os pedidos sem data de entrega estão cancelados.")
```

# BLOCO 4 FORMATAÇÃO TEMPORAL - DATETIME

```python
from datetime import datetime

# Função para processar e validar cada data individualmente

def processar_data_venda(val):
    if pd.isna(val):
        return None, None, None, None, "Invalida", False
    
    val_str = str(val).strip()
    
    try:
        # Tenta converter a string no formato esperado YYYY-MM-DD HH:MM:SS

        dt = datetime.strptime(val_str, "%Y-%m-%d %H:%M:%S")
        # Cria a string formatada no padrão brasileiro DD/MM/YYYY

        data_br = dt.strftime("%d/%m/%Y")
        return dt.year, dt.month, dt.day, dt.hour, data_br, True
    except Exception:
        # Caso a data venha com formato errado (ex: 'data_invalida'), não trava o código

        return None, None, None, None, "Invalida", False

# Aplicamos a função linha a linha

resultados_data = df['data_venda'].apply(processar_data_venda)

# Extraímos os valores retornados para novas colunas do DataFrame

df['ano'] = [r[0] for r in resultados_data]
df['mes'] = [r[1] for r in resultados_data]
df['dia'] = [r[2] for r in resultados_data]
df['hora'] = [r[3] for r in resultados_data]
df['data_venda_br'] = [r[4] for r in resultados_data]
df['data_venda_valida'] = [r[5] for r in resultados_data]

datas_invalidas = (~df['data_venda_valida']).sum()

print("=== BLOCO 4: PROCESSAMENTO TEMPORAL (DATETIME) ===")
print(f"1. Datas de venda válidas: {df['data_venda_valida'].sum()}")
print(f"2. Datas de venda inválidas ou vazias: {datas_invalidas}")
print("\nExemplo de registros processados (5 primeiras linhas):")
display(df[['data_venda', 'data_venda_br', 'ano', 'mes', 'hora', 'data_venda_valida']].head(5))
```
