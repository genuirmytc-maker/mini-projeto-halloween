```python
caminho = "halloween_vendas.csv"
df = pd.read_csv(caminho, sep=",", encoding="utf-8", dtype=str)
```

```python
df 
```

Preencha categoria vazia com a string "sem categoria"

```python
df['categoria']
```

```python
vazios = df['categoria'].isna() | df['categoria'].str.strip().eq('')

df.loc[vazios]
```

```python
df.loc[vazios, 'categoria'] = 'sem categoria'
```

Verifique quantidade (inteiro
maior que zero) e preco_unitario (preenchido e não negativo)

```python
quantidade = pd.to_numeric(df['quantidade'], errors='coerce')
preco = pd.to_numeric(df['preco_unitario'], errors='coerce')

validos = (
    quantidade.notna()
    & (quantidade > 0)
    & (quantidade % 1 == 0)
    & preco.notna()
    & (preco >= 0)
)

print(f"Registros válidos: {validos.sum()}")
```
